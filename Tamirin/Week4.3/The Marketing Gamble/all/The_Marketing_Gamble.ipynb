```python
#TO-DO
train_df = pd.read_csv("train.csv")
test_df = pd.read_csv("test.csv")

train_df
```

<h2 align=right style="line-height:200%;font-family:vazir;color:#0099cc">
<font face="vazir" color="#0099cc">
پیش‌پردازش و مهندسی ویژگی
</font>
</h2>


<p dir=rtl style="direction: rtl;text-align: right;line-height:200%;font-family:vazir;font-size:medium">
<font face="vazir" size=3>
    در این سوال شما می‌توانید از هر تکنیک پیش‌پردازش/مهندسی ویژگی که آموختید، استفاده کنید.
    <br>
    تکنیک‌هایی که استفاده می‌کنید به شکل مستقیم مورد ارزیابی توسط سامانه داوری قرار <b>نمی‌گیرند.</b> بلکه همه آن‌ها در دقت مدل شما تاثیر خواهند گذاشت؛ بنابراین هر چه پیش‌پردازش/مهندسی ویژگی بهتری انجام دهید تا دقت مدل بهبود پیدا کند، امتیاز بیشتری از این سوال کسب خواهید کرد.
</font>
</p>

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder

# Removing duratino
train_df = train_df.drop(columns=["duration"], errors="ignore")
test_df = test_df.drop(columns=["duration"], errors="ignore")


# Ordinal encoder
education_order = [
    "illiterate",
    "basic.4y",
    "basic.6y",
    "basic.9y",
    "high.school",
    "professional.course",
    "university.degree",
]
month_order = ["mar", "apr", "may", "jun", "jul", "aug", "sep", "oct", "nov", "dec"]
day_order = ["mon", "tue", "wed", "thu", "fri"]
ordinal_columns = ["education", "month", "day_of_week"]

ord_encoder = OrdinalEncoder(
    categories=[education_order, month_order, day_order],
    handle_unknown="use_encoded_value",
    unknown_value=-1,
)


train_df[ordinal_columns] = ord_encoder.fit_transform(train_df[ordinal_columns])
test_df[ordinal_columns] = ord_encoder.transform(test_df[ordinal_columns])


# Onehot encoder.
onehot_columns = ["loan", "marital", "contact", "job"]
one_encoder = OneHotEncoder(
    drop="first",
    sparse_output=False,
    handle_unknown="ignore",
).set_output(transform="pandas")

train_one = one_encoder.fit_transform(train_df[onehot_columns])
test_one = one_encoder.transform(test_df[onehot_columns])

train_df = train_df.drop(columns=onehot_columns).join(train_one)
test_df = test_df.drop(columns=onehot_columns).join(test_one)

# housing and default
map_or_not: dict = {
    "unknown" : 0,
    "no" : 1,
    "yes" : 2
}
train_df["housing"] = train_df["housing"].map(map_or_not)
test_df["housing"] = test_df["housing"].map(map_or_not)
train_df["default"] = train_df["default"].map(map_or_not)
test_df["default"] = test_df["default"].map(map_or_not)

train_df.loc[train_df["pdays"] == 999, "pdays"] = 0
test_df.loc[test_df["pdays"] == 999, "pdays"] = 0

poutcome_or_not: dict = {
    "nonexistent" : 0,
    "success" : 2,
    "failure" : 1
}
train_df["poutcome"] = train_df["poutcome"].map(map_or_not)
test_df["poutcome"] = test_df["poutcome"].map(map_or_not)

train_df["was_contacted"] = (train_df["pdays"] != 999).astype(int)
test_df["was_contacted"] = (test_df["pdays"] != 999).astype(int)

train_df["prev_success"] = (train_df["poutcome"] == "success").astype(int)
train_df["prev_contacted"] = (train_df["poutcome"] != "nonexistent").astype(int)
test_df["prev_success"] = (test_df["poutcome"] == "success").astype(int)
test_df["prev_contacted"] = (test_df["poutcome"] != "nonexistent").astype(int)
```

```python
# getting the most important features
from sklearn.tree import  DecisionTreeClassifier
model = DecisionTreeClassifier(random_state=42)
train_x = train_df.drop(columns=["y"])
train_y = train_df["y"]
test_x = test_df

model.fit(train_x, train_y)
feature_importance = pd.DataFrame(
    {
        "feature name" : model.feature_names_in_,
        "feature importance" : model.feature_importances_,

    }
).sort_values("feature importance", ascending = False)
feature_importance
```

```python
from sklearn.feature_selection import RFECV
from sklearn.tree import  DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.model_selection import  StratifiedKFold

train_x = train_df.drop(columns=["y"])
train_y = train_df["y"]
test_x = test_df

estimator = DecisionTreeClassifier()
cv = StratifiedKFold(shuffle=True, random_state=42, n_splits=5)
rfc_finder = RFECV(estimator=estimator, min_features_to_select=12, cv=cv)
rfc_finder.fit(train_x, train_y)
```

```python
from sklearn.preprocessing import  PolynomialFeatures, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.model_selection import  train_test_split

chosen_feature: list = [
    "age",
    "education",
    "default",
    "housing",
    "month",
    "day_of_week",
    "campaign",
    "pdays",
    "previous",
    "cons_price_idx",
    "cons_conf_idx",
    "euribor3m",
    "nr_employed",
    "loan_yes",
    "marital_married",
    "marital_single",
    "contact_telephone",
    "job_blue-collar",
    "job_entrepreneur",
    "job_housemaid",
    "job_management",
    "job_retired",
    "job_self-employed",
    "job_services",
    "job_student",
    "job_technician",
    "job_unemployed",
]

train_x, validation_x, train_y, validation_y = train_test_split(
    train_x, train_y, test_size=0.2, stratify=train_y, random_state=42
)

train_x = train_x[chosen_feature]
test_x = test_x[chosen_feature]
validation_x = validation_x[chosen_feature]

scaler = StandardScaler()


train_x = scaler.fit_transform(train_x)
test_x = scaler.transform(test_x)
validation_x = scaler.transform(validation_x)
```

```python
from sklearn.model_selection import  GridSearchCV, StratifiedKFold
from sklearn.tree import DecisionTreeClassifier

cv = StratifiedKFold(random_state=42, shuffle=True, n_splits=5)
grid_params = {
    "max_depth" : [1, 5, 10, 15],
    "min_samples_leaf" : [1, 5, 10, 20],
    "min_samples_split" : [2, 5, 10, 20, 40],
    "criterion" : ["gini", "log_loss", "entropy"],
    "splitter" : ["random", "best"]
}   
grid_search = GridSearchCV(param_grid=grid_params, cv=cv, estimator=DecisionTreeClassifier(random_state=42))
grid_search.fit(train_x, train_y)
```

    معیاری که برای ارزیابی عملکرد مدل انتخاب کرده‌ایم، <code>f1_score</code> نام دارد.
    <br>
    این معیار، سنجه ارزیابی کیفیت مدل شماست. به عبارت بهتر در سامانه داوری هم از همین معیار برای نمره‌دهی استفاده شده است.

```python
# evaluate model
from sklearn.metrics import f1_score
probabilities = model.predict_proba(validation_x)[:, 1]
y_prediction = model.predict(validation_x)

best_thresh = 0.5
best_f1 = 0
for thresh in np.arange(0.2, 0.7, 0.05):
    preds = (probabilities >= thresh).astype(int)
    score = f1_score(validation_y, preds)
    if score > best_f1:
        best_f1 = score
        best_thresh = thresh
print(best_f1)
print(best_thresh)
```
